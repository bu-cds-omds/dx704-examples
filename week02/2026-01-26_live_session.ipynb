```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(2026_01_26)
```

```python
np.sqrt(np.arange(100, dtype=np.float64) + 1.0).astype(np.int64)
```

```python
daily = np.random.binomial(np.pow(np.arange(100, dtype=np.float64) + 1.0, 0.6).astype(np.int64), 0.5, 100)
```

```python
daily.sum()
```

```python
plt.plot(daily)
plt.title("Widgets Sold Per Day")
plt.xlim(0)
plt.ylim(0)
plt.plot();
```

```python
plt.plot(np.cumsum(daily))
plt.title("Cumulative Widgets Sold")
plt.xlim(0)
plt.ylim(0)
plt.plot();
```

```python
plt.plot(500 - np.cumsum(daily), color="C0", label="Buy up front")
plt.title("Remaining Inventory")
plt.legend()
plt.xlim(0)
plt.ylim(0)
plt.plot();
```

```python
plt.plot((np.arange(100) + 1) * 5 - np.cumsum(daily), color="C1", label="Build 5/day")
plt.title("Remaining Inventory")
plt.legend()
plt.xlim(0)
plt.ylim(0)
plt.plot();
```

```python
plt.plot((np.arange(100) + 1) * 4 - np.cumsum(daily), color="C2", label="Build 4/day")
plt.title("Remaining Inventory")
plt.legend()
plt.xlim(0)
plt.axhline(0, color="grey")
plt.plot();
```

```python
plt.plot(np.cumsum(daily)*10)
plt.title("Cumulative Revenue")
plt.xlim(0)
plt.ylim(0)
plt.plot();
```
