```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
poem_original = """Sun sinks, sinks into the sea,
Light fades, fades so tenderly.
The waves repeat what day has done—
Goodbye, goodbye, departing sun."""
print(poem_original)
```

```python
poem_clean = ''.join(c for c in poem_original if c.isalpha() or c.isspace()).lower()
print(poem_clean)
```

```python
poem_list = poem_clean.split()
poem_words = set(poem_list)
poem_words
```

```python
poem_word_counts = {word: poem_list.count(word) for word in sorted(poem_words)}
poem_word_counts
```

```python
xs = np.linspace(-8, 8, 1000)
plt.plot(xs, 1 / (1 + np.exp(-xs)))
plt.title("Standard Logistic Function")
plt.ylim(0,1)

plt.show();
```
