```python
def train_ibm_model_0(lang1_name, lang2_name, corpus, iterations=15, font_name='sans-serif'):
    # Tokenize corpus
    tokenized_corpus = []
    lang1_vocab_set = set()
    lang2_vocab_set = set()

    for sent1, sent2 in corpus:
        words1 = sent1.strip().split()
        words2 = sent2.strip().split()
        tokenized_corpus.append((words1, words2))
        lang1_vocab_set.update(words1)
        lang2_vocab_set.update(words2)

    lang1_vocab = sorted(list(lang1_vocab_set))
    lang2_vocab = sorted(list(lang2_vocab_set))

    l1_to_idx = {w: i for i, w in enumerate(lang1_vocab)}
    l2_to_idx = {w: i for i, w in enumerate(lang2_vocab)}

    V1 = len(lang1_vocab)
    V2 = len(lang2_vocab)

    # Initialize translation probabilities t(f|e) uniformly
    t = np.ones((V1, V2)) / V2

    history = [t.copy()]

    for _ in range(iterations):
        # Initialize counts
        count_f_given_e = np.zeros((V1, V2))
        total_e = np.zeros(V1)

        for words1, words2 in tokenized_corpus:
            # Calculate normalization denominator
            s_total = {}
            for f in words2:
                s_total[f] = sum(t[l1_to_idx[e], l2_to_idx[f]] for e in words1)

            # Collect fractional counts
            for f in words2:
                f_idx = l2_to_idx[f]
                denom = s_total[f]
                if denom > 0:
                    for e in words1:
                        e_idx = l1_to_idx[e]
                        val = t[e_idx, f_idx] / denom
                        count_f_given_e[e_idx, f_idx] += val
                        total_e[e_idx] += val

        # Re-estimate probabilities
        for e_idx in range(V1):
            denom = total_e[e_idx]
            if denom > 0:
                t[e_idx, :] = count_f_given_e[e_idx, :] / denom
            else:
                t[e_idx, :] = 1.0 / V2

        history.append(t.copy())

    # Set up the animation with the explicit font selection
    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(history[0], cmap='Blues', aspect='auto', vmin=0, vmax=1)
    fig.colorbar(im, ax=ax, label='Probability')

    ax.set_xticks(range(V2))
    ax.set_xticklabels(lang2_vocab, rotation=45, ha='right', fontfamily=font_name)
    ax.set_yticks(range(V1))
    ax.set_yticklabels(lang1_vocab, fontfamily=font_name)
    ax.set_title(f"Translation Probability P({lang2_name} | {lang1_name}) - Init", fontfamily=font_name)
    fig.tight_layout()

    def update(frame):
        im.set_array(history[frame])
        ax.set_title(f"Translation Probability P({lang2_name} | {lang1_name}) - Iteration {frame}", fontfamily=font_name)
        return [im]

    anim = animation.FuncAnimation(fig, update, frames=len(history), interval=500, blit=True)
    plt.close(fig)

    return lang1_vocab, lang2_vocab, t, anim

# Test French with English corpus
lang1_vocab, lang2_vocab, final_matrix, anim = train_ibm_model_0(
    "English", "French", english_french_corpus, iterations=15
)

# Display translation table nicely
df = pd.DataFrame(final_matrix, index=lang1_vocab, columns=lang2_vocab)
print("Final Translation Probability Matrix:")
display(df.round(3))

# Render animation!
HTML(anim.to_jshtml())
```
